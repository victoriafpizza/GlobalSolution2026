# Embeddings, FAISS e busca
Execute com o kernel local a partir da raiz da disciplina. As avaliações históricas permanecem preservadas.

```python
import json
from pathlib import Path
from tempfile import TemporaryDirectory
from rag.embeddings import Embeddings, MODEL, REVISION, DIM
from rag.indexador import construir
from rag.busca import Busca
print(MODEL, REVISION, DIM)
encoder = Embeddings()
with TemporaryDirectory() as temp:
    path = Path(temp) / 'indice.zip'
    print('Vetores construídos:', construir(encoder, path))
    for item in Busca(encoder, path).consultar('Em que espaços podem existir barreiras urbanísticas?'):
        print(item['chunk_id'], round(item['score'], 4), item['secao_artigo'])
historico = json.loads(Path('avaliacao/resultados_recuperacao.json').read_text(encoding='utf-8'))
print('Recall histórico:', historico['ranking']['factual'])
```
