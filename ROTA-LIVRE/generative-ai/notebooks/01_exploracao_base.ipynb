# Exploração e integridade da base
Execute com o kernel local a partir da raiz da disciplina. As avaliações históricas permanecem preservadas.

```python
from collections import Counter
from scripts.validar_base import validar_base, validar_casos
fontes, chunks = validar_base()
casos = validar_casos(fontes, chunks)
for key, fonte in fontes.items():
    print(key, fonte['titulo'], fonte['url'])
print('Chunks por fonte:', dict(Counter(c['source_id'] for c in chunks.values())))
print('Casos por categoria:', dict(Counter(c['categoria'] for c in casos.values())))
```
